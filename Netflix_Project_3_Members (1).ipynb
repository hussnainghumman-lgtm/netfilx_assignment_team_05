## 3. Inspect the datasets
We check the first rows, column names, data types, and summary information.

```python
print("USERS")
display(users.head())
users.info()

print("\\nMOVIES")
display(movies.head())
movies.info()

print("\\nWATCH HISTORY")
display(watch_history.head())
watch_history.info()
```

## 4. Data cleaning
We check missing values, duplicate rows, and invalid links between tables. The sample data is designed to be mostly complete, but we still perform the checks.

```python
print("Missing values in users:")
display(users.isnull().sum())

print("Missing values in movies:")
display(movies.isnull().sum())

print("Missing values in watch history:")
display(watch_history.isnull().sum())

print("Duplicate rows:")
print("Users:", users.duplicated().sum())
print("Movies:", movies.duplicated().sum())
print("Watch history:", watch_history.duplicated().sum())

# Remove duplicate rows if any
users = users.drop_duplicates()
movies = movies.drop_duplicates()
watch_history = watch_history.drop_duplicates()

# Convert watch_date to a date/time data type
watch_history["watch_date"] = pd.to_datetime(watch_history["watch_date"])

# Check that every watch record refers to an existing user and movie
print("Unknown user IDs:", (~watch_history["user_id"].isin(users["user_id"])).sum())
print("Unknown movie IDs:", (~watch_history["movie_id"].isin(movies["movie_id"])).sum())
```

## 5. Exploratory data analysis (EDA)
We combine the viewing records with movie and user details to answer simple questions about viewing behaviour.

```python
watch_details = watch_history.merge(movies, on="movie_id", how="left")
watch_details = watch_details.merge(users, on="user_id", how="left")

display(watch_details.head())
print("Total watch records:", len(watch_details))
print("Total watch minutes:", watch_details["watch_minutes"].sum())
print("Average watch minutes:", round(watch_details["watch_minutes"].mean(), 2))
print("Completed watches:", watch_details["completed"].sum())
```

## 6. Visualizations
The following charts show viewing activity by genre, subscription plan, and movie completion.

```python
# Chart 1: Number of watch records by genre
genre_counts = watch_details["genre"].value_counts()
genre_counts.plot(kind="bar", title="Watch Records by Movie Genre")
plt.xlabel("Genre")
plt.ylabel("Number of watch records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Number of users by subscription plan
subscription_counts = users["subscription"].value_counts()
subscription_counts.plot(kind="bar", title="Users by Subscription Plan")
plt.xlabel("Subscription plan")
plt.ylabel("Number of users")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Chart 3: Completed vs not completed watches
completion_counts = watch_details["completed"].value_counts().rename(
    index={True: "Completed", False: "Not completed"}
)
completion_counts.plot(kind="bar", title="Completed vs Not Completed Watches")
plt.xlabel("Viewing status")
plt.ylabel("Number of watch records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. Business insights
Use the actual results printed below to support the conclusions. Because the dataset contains only 10 viewing records, these are practice observations and should not be generalized to all Netflix customers.

```python
print("Most watched genre:")
print(watch_details["genre"].value_counts().head(1))

print("\\nUsers by subscription:")
display(users["subscription"].value_counts())

completion_rate = watch_details["completed"].mean() * 100
print(f"Completion rate: {completion_rate:.1f}%")

print("\\nViewing minutes by country:")
display(watch_details.groupby("country")["watch_minutes"].sum().sort_values(ascending=False))

print("\\nViewing records by movie:")
display(watch_details.groupby("title")["watch_id"].count().sort_values(ascending=False))
```

### Example business insights
1. The genre chart identifies which genres appear most often in this sample. Netflix could use genre popularity as one input when planning recommendations.
2. The subscription chart shows the mix of Basic, Standard, and Premium users in this sample. A real business analysis could compare viewing activity and retention across plans.
3. The completion chart shows how many viewing records were completed. A larger real dataset could help identify movies that viewers often stop watching early.

**Limitation:** Each user has only one watch-history record here. The sample is too small to make reliable claims about customer behaviour, revenue, churn, or the overall Netflix audience.
